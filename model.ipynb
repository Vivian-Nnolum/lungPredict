### Check for duplicates

```python
df = df.drop_duplicates()
```

```python
from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()

df["GENDER"] = encode.fit_transform(df['GENDER'])
df['LUNG_CANCER'] = encode.fit_transform(df['LUNG_CANCER'])
```

```python
df.head()
```

```python
df['SMOKING'] = df['SMOKING'].map({
    1:0,2:1
})

df['YELLOW_FINGERS'] = df['YELLOW_FINGERS'].map({
    1:0,2:1
})

df['ANXIETY'] = df['ANXIETY'].map({1:0,2:1})
df['PEER_PRESSURE'] = df['PEER_PRESSURE'].map({1:0,2:1})
df['CHRONIC DISEASE'] = df['CHRONIC DISEASE'].map({1:0,2:1})
df['FATIGUE '] = df['FATIGUE '].map({1:0,2:1})
df['ALLERGY'] = df['ALLERGY '].map({1:0,2:1})
df['WHEEZING'] = df['WHEEZING'].map({1:0,2:1})
df['ALCOHOL CONSUMING'] = df['ALCOHOL CONSUMING'].map({1:0,2:1})
df['COUGHING'] = df['COUGHING'].map({1:0,2:1})
df['SHORTNESS OF BREATH'] = df['SHORTNESS OF BREATH'].map({1:0,2:1})
df['SWALLOWING DIFFICULTY'] = df['SWALLOWING DIFFICULTY'].map({1:0,2:1})
df['CHEST PAIN'] = df['CHEST PAIN'].map({1:0,2:1})
```

#### Spliting Data

```python
from sklearn.model_selection import train_test_split


x = df.drop(["LUNG_CANCER"],axis=1)
y = df['LUNG_CANCER']
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,stratify=y,random_state=42)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE()

x_train_s,y_train_s = smote.fit_resample(x_train,y_train)
```

```python
y_train_s.value_counts()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,recall_score,precision_score,classification_report

model = XGBClassifier(n_estimators=200,learning_rate=0.8,subsample=0.8,max_depth=3,random_state=12)

model.fit(x_train_s,y_train_s)

pred = model.predict(x_test)

acc = accuracy_score(y_test,pred)
rec = recall_score(y_test,pred)
pre = precision_score(y_test,pred)

print(f"Accuracy Score: {acc}\n Recall Score: {rec} \n Precision Score: {pre}")
print(classification_report(y_test,pred))
```

```python
import optuna
from xgboost import XGBClassifier
from sklearn.metrics import recall_score
from sklearn.model_selection import train_test_split

# Objective function for Optuna
def objective(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 300, 1000),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 1),
        "max_depth": trial.suggest_int("max_depth", 0, 20),
        "subsample": trial.suggest_float("subsample", 0.2, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.2, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 0, 20),
        "gamma": trial.suggest_float("gamma", 0, 20),
        "reg_alpha": trial.suggest_float("reg_alpha", 0, 20),
        "reg_lambda": trial.suggest_float("reg_lambda", 0, 20),
        "random_state": 42,
        "n_jobs": -1,
        "scale_pos_weight": (y_train_s == 0).sum() / (y_train_s == 1).sum(),
        "eval_metric": "logloss"
    }
    
    # THRESHOLD = 0.25
    
    model = XGBClassifier(**params)
    model.fit(x_train_s, y_train_s)
    proba = model.predict_proba(x_test)[:, 1]
    # y_pred = (proba >= THRESHOLD).astype(int)  # Keep your tuned threshold
    y_pred = model.predict(x_test)
    print(f"Accuracy Score: {accuracy_score(y_test,y_pred)}")
    return recall_score(y_test, y_pred)  # Optimize recall for churners

# Run Optuna
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=30)

print("Best Params:", study.best_params)
print("Best Recall:", study.best_value)

params = study.best_params
```

```python
best_model = XGBClassifier(n_estimators= 634,learning_rate= 0.0834406171723694,max_depth= 19,subsample= 0.9560301670686766,colsample_bytree= 0.5610548250635559,min_child_weight= 18,gamma= 15.437091403894819, reg_alpha= 2.253803670031582,reg_lambda= 11.129358439611085)

# best_model = XGBClassifier(**params)

best_model.fit(x_train_s,y_train_s)

pred = best_model.predict(x_test)

acc = accuracy_score(y_test,pred)
rec = recall_score(y_test,pred)
pre = precision_score(y_test,pred)

print(f"Accuracy Score: {acc}\n Recall Score: {rec} \n Precision Score: {pre}")
print(classification_report(y_test,pred))
```
